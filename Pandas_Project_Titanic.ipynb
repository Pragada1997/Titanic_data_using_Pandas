```python
#Using dataset titanic.dataset.csv.
import numpy as np
import pandas as pd
data=pd.read_csv("titanic_dataset.csv")
data.head()
```

```python
#Calculating the total of count of male and female.
head_count=data.groupby(["Sex"])["PassengerId"].count()
print("Total Number of Males and Females are:", head_count)
```

```python
#Finding all the null values
data.isnull().sum()
```

```python
new_data=data.ffill()  #used forward fill method to handle the null values
new_data
```

```python
#replaced one null value in Cabin column
new_data.fillna({
    'Cabin':'C85'
})
```

```python
#Filtering the data again by removing the null values if any
new_data.dropna(inplace=True)
new_data
```

```python
# Checking if there is any null value present in the data
new_data.isnull().sum()
```

```python
#removing all unwanted columns from the data
new_data.drop(['Embarked','Cabin','SibSp','Ticket'],axis=1,inplace=True)
new_data
```

```python
#Find overall survival rate
Survival_rate=new_data["Survived"].mean()
print("Overall Suvived rate:",Survival_rate)
```

```python
#Cross-validate the female and male data.
Total_count=new_data["PassengerId"].count()
print("Total Paasengers after removing null values: ",Total_count)

head_count=new_data.groupby(["Sex"])["PassengerId"].count()
print("Total Number of Males and Females are:")
head_count
```

```python
#Find if there is any correlation between passenger class and survival
co_relation=new_data["Pclass"].corr(new_data['Survived'])
print("Correlation between Passenger class and Survivals:",co_relation)
```
